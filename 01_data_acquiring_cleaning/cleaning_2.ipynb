```python
def clean_text(df, col):
    import re
    import emoji

    df[col] = df[col].str.lower()
    df[col] = df[col].str.strip()

    # Remove URLs
    df[col] = df[col].apply(lambda x: re.sub(r'https?://\S+|www\.\S+', '', x))

    # Replace tickers and hashtags with placeholder tokens
    df[col] = df[col].apply(lambda x: re.sub(r'\$[A-Za-z]+', '<TICKER>', x))
    df[col] = df[col].apply(lambda x: re.sub(r'\#[A-Za-z]+', '<HASHTAG>', x))

    # Remove backtick-style quotes
    df[col] = df[col].apply(lambda x: re.sub(r"``\s*(.*?)\s*''", r'\1', x))

    # Emoji handling
    df[col] = df[col].apply(lambda x: emoji.demojize(x))

    # Normalize ellipses (longest pattern first)
    df[col] = (df[col]
        .str.replace(".....…", ".", regex=False)
        .str.replace("…", ".", regex=False)
        .str.replace("...", ".", regex=False)
        .str.replace("..", ".", regex=False)
    )

    # Remove @mentions
    df[col] = df[col].apply(lambda x: re.sub(r'(@[A-Za-z0-9]*)', '', x))

    # Remove stray smart quotes/em-dash
    df[col] = df[col].apply(lambda x: re.sub(r'[""''—]', '', x))

    # Remove mojibake / non-ASCII leftover garbage
    df[col] = df[col].apply(lambda x: re.sub(r'[^\x00-\x7F]+', '', x))

    return df
```

```python
df = df.dropna()
```

```python
df = clean_text(df,'text')
```

```python
df['text'] = df['text'].str.replace('{','').str.replace('}','')
df['text'] = df['text'].str.replace('[','').str.replace(']','')
```

```python
df['text'] = df['text'].str.replace('\n',' ')
df['text'] = df['text'].str.replace('/n',' ')
```

```python
df.to_csv('Modelling_Ready.csv')
```
