# Converting each data into data frame and then combining them

```python
 # Financial data
df_fin = pd.DataFrame(phrasebank['train'])
df_fin_test = pd.DataFrame(phrasebank['test'])
df_fin
```

```python
 # Financial data
df_twe = pd.DataFrame(twitter_news['train'])
df_twe_test = pd.DataFrame(twitter_news['validation'])
df_twe
#  0 - Negative
# 1 - positive
# 2 - neutral
```

```python
# Since the data set is converted into data from successfully now refurnishing the labels
```

```python
df_fin[['label','label_text']].value_counts()
# for financial data -
# 0 = negative ,
# 1 = neutral,
# 2 = positive
```

```python
# Assigning same labels to twitter data

df_twe['label'].value_counts()
df_twe['label'] = df_twe['label'].map({0 : 0 ,
                                       1 : 2,
                                       2 : 1})
df_twe['label'].value_counts()
```

```python
# Doing same for test data -

df_twe_test['label'].value_counts()
df_twe_test['label'] = df_twe_test['label'].map({0 : 0 ,
                                       1 : 2,
                                       2 : 1})
df_twe_test['label'].value_counts()
```

```python
# Combining the financial data including test and train
df1 = pd.concat([df_fin,df_fin_test],axis=0)
df1.sample()
```

```python
df2 = pd.concat([df_twe,df_twe_test],axis=0)
df2.sample()
```

```python
df2['label_text'] = df2['label'].map({1: 'neutral',
                                      2 : 'positive',
                                      0 : 'negative'})
df2.columns
df2 = df2[['text', 'label_text', 'label']]
```

```python
df1 = df1.reset_index(drop=True)
df2 = df2.reset_index(drop=True)
```

```python
df1.to_csv('Financial_sentiment_data.csv')
df2.to_csv('Financial_tweets_sentiment_data.csv')
```

```python
# Combining both the data
df = pd.concat([df1,df2],axis=0).reset_index(drop = True)
idx = df.index
# shuffling the data
```

```python
np.random.seed(45)
idx = np.random.permutation(idx)
df = df.loc[idx,:].reset_index(drop = True)
```

```python
df.to_csv('Combined_financial_sentiment_analysis_data.csv')
```
