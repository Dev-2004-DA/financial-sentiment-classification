```python
# Data cleaning pipeline
def clean_text(df, col ):

  """lowercasing"""
  df[col] = df[col].str.lower()

  """stripping"""
  df[col] = df[col].str.strip()

  """ Removing HTML tag if any"""
  from bs4 import BeautifulSoup
  df[col] = df[col].apply(
                          lambda x: BeautifulSoup(str(x), 'html.parser').get_text()
                          )
  import re
  df[col] = df[col].apply(
                        lambda x : re.sub(r'https?://\S+|www\.\S+', '', x)
                    )
  """Removing ticker"""

  df[col] = df[col].apply(
      lambda x: re.sub(r'\$[A-Za-z]+','<TICKER>',x))

  """Remove hashtags"""
  df[col] = df[col].apply(
      lambda x: re.sub(r'\#[A-Za-z]+','<Hashtag>',x))

  """Removing extra commas,inverted commas"""
  df[col] = df[col].apply(
      lambda x : re.sub(r"``\s*(.*?)\s*''",r'\1',x))

  """Spell correction"""
  from textblob import TextBlob
  df[col] = df[col].apply(
      lambda x: TextBlob(x).correct().string)

  """Emojis Handling"""
  import emoji
  df[col] = df[col].apply(
      lambda x : emoji.demojize(x)
  )

  return df
```

# Data Cleaning is Done

```python
# Data cleaning pipeline
def clean_text(df, col ):

  """lowercasing"""
  df[col] = df[col].str.lower()

  """stripping"""
  df[col] = df[col].str.strip()

  import re
  df[col] = df[col].apply(
                        lambda x : re.sub(r'https?://\S+|www\.\S+', '', x)
                    )
  """Removing ticker"""

  df[col] = df[col].apply(
      lambda x: re.sub(r'\$[A-Za-z]+','<TICKER>',x))

  """Remove hashtags"""
  df[col] = df[col].apply(
      lambda x: re.sub(r'\#[A-Za-z]+','<Hashtag>',x))

  """Removing extra commas,inverted commas"""
  df[col] = df[col].apply(
      lambda x : re.sub(r"``\s*(.*?)\s*''",r'\1',x))

  """Emojis Handling"""
  import emoji
  df[col] = df[col].apply(
      lambda x : emoji.demojize(x)
  )

  df[col] = df[col].str.replace("...",".").str.replace("..",".").str.replace(".....…",".").str.replace("…",".")

  """Removing @"""
  df[col] = df[col].apply(
      lambda x : re.sub(r'(@[A-Za-z0-9]*)','',x)
  )

  # Removing ascii charcaters
  df[col] = df[col].apply(
      lambda x : re.sub('[""''—]','',x)
  )

  return df
```
